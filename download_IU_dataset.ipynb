```python
from pathlib import Path
import urllib.request
import tarfile
import os

# Không đụng tới file NLMCXR_png.tgz cũ nữa
download_dir = Path("iu_download_v2")
raw_dir = Path("data_iu/raw_v2")

download_dir.mkdir(parents=True, exist_ok=True)
raw_dir.mkdir(parents=True, exist_ok=True)

files = {
    download_dir / "NLMCXR_png_v2.tgz": "https://openi.nlm.nih.gov/imgs/collections/NLMCXR_png.tgz",
    download_dir / "NLMCXR_reports_v2.tgz": "https://openi.nlm.nih.gov/imgs/collections/NLMCXR_reports.tgz",
}

def download_file(url, output_path):
    output_path = Path(output_path)
    temp_path = Path(str(output_path) + ".part")

    # Chỉ xóa file .part trong folder mới, không đụng file cũ
    if temp_path.exists():
        temp_path.unlink()

    print(f"⬇️ Đang tải: {output_path}")

    def progress(block_num, block_size, total_size):
        downloaded = block_num * block_size
        if total_size > 0:
            percent = min(downloaded * 100 / total_size, 100)
            print(f"\r{percent:.2f}% downloaded", end="")

    urllib.request.urlretrieve(url, temp_path, progress)
    temp_path.rename(output_path)

    print(f"\n✅ Tải xong: {output_path}")
    print(f"Size: {output_path.stat().st_size / (1024**3):.2f} GB")

def test_tgz(path):
    print(f"🔍 Kiểm tra file nén: {path}")
    with tarfile.open(path, "r:gz") as tar:
        count = 0
        for _ in tar:
            count += 1
    print(f"✅ File OK, có {count} items")

# Download lại bằng tên mới
for path, url in files.items():
    download_file(url, path)

# Kiểm tra file có bị corrupt không
for path in files.keys():
    test_tgz(path)

# Extract sang folder mới
for path in files.keys():
    print(f"📦 Đang giải nén: {path}")
    with tarfile.open(path, "r:gz") as tar:
        tar.extractall(raw_dir)
    print(f"✅ Giải nén xong: {path}")

print("\n🎉 Hoàn tất!")
print("Dữ liệu mới nằm ở:", raw_dir.resolve())
print("Một vài file/folder bên trong:")
print(os.listdir(raw_dir)[:20])
```
